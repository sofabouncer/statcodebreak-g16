```python
#we make a string of letters for the alphabet
alphabet = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ ' 

#creating a decrypting function
def  decrypt (ciphertext, key):
    plaintext = ''
    #for each letter in our encrypted word we find the index, subtract the key and then find this new indexed letter
    for letter in ciphertext:
        index = alphabet.index(letter)
        new_index = index - key
        if new_index < 0:
            new_index += 27
        plaintext += alphabet[new_index]
    return plaintext

#Now use a for loop to go through every possible combination and print the results
for i in range (27):
    cipher_1 = decrypt (ciphertexts[1], i)
    print(f"Shift {i:2d} | {cipher_1[:70]}...")

#Define our correct cipher key
cipher_correct = decrypt (ciphertexts [1], 18)

print()
print ("Hence we can see that the key = 18 and our decrypted message is")
print()
print (cipher_correct) 
    
```

```python
#We can now apply this method to all 12 ciphers
# Store all decoded first guesses
first_guesses = {}

# Loop through every ciphered message
for cipher in ciphertexts:
    secret_text = ciphertexts[cipher]
    
    #Count character frequencies for THIS specific ciphertext
    cipher_char_count = []
    for char in allowed_chars:
        cipher_char_count.append(secret_text.count(char))
        
    #Pair and sort from highest to lowest frequency
    paired_cipher_list = list(zip(cipher_char_count, allowed_chars))
    paired_cipher_list.sort(reverse=True)
    
    # Build a custom translation dictionary for this message
    cipher_dict = {}
    for k in range(27):
        cipher_char = paired_cipher_list[k][1]
        english_char = paired_list[k][1]
        cipher_dict[cipher_char] = english_char
        
    # Decode the message using its dictionary
    decoded_chars = []
    for char in secret_text:
        decoded_chars.append(cipher_dict[char])
        
    first_guess = "".join(decoded_chars)
    first_guesses[cipher] = first_guess
    
    #Print a clean, spaced-out result for each message
    print(f"=== Ciphertext {cipher} ===")
    print(first_guess[:200] + "...")  # Prints the first 200 characters
    print()
    
```
