```python

# Define the alphabet containing 26 letters and a space
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

# Function to decrypt a message using a given shift
def caesar_decrypt(message, shift):
    decrypted = ""

    for char in message:
        position = alphabet.index(char)
        new_position = (position - shift) % 27
        decrypted += alphabet[new_position]

    return decrypted
```

### Mathematical Explanation

A Caesar cipher is a substitution cipher in which every character is shifted by a fixed number of positions in the alphabet.

Since our alphabet contains 26 letters and a space, there are 27 possible shifts.

The decryption formula is:

$$
D(x) = (x-k) \bmod 27
$$

where:

- $x$ is the position of the encrypted character.
- $k$ is the shift value.
- $D(x)$ is the position of the decrypted character.

The modulo operation ensures that character positions wrap around the alphabet.

By testing all 27 possible shifts, the correct message can be identified without knowing the encryption key beforehand. This method is known as brute-force decryption.

```python

# The correct shift identified by brute force
correct_shift = 18

# Decrypt the entire first message
decoded_message = caesar_decrypt(ciphertexts[1], correct_shift)

print("Correct shift:", correct_shift)
print()
print("Decrypted message:")
print(decoded_message)
```

Although brute-force decryption is effective for this simple cipher, it becomes impractical for more complex substitution ciphers, which have a much larger number of possible character mappings. Therefore, statistical methods will be explored in the following sections to decrypt more challenging messages.

## 3. Frequency Analysis of Substitution Ciphers

### Aim

The aim is to estimate substitution keys for Messages 2–12 using character frequencies.

### Method

We count how often each of the 26 letters and the space character appears in the cleaned *Moby Dick* text. We then do the same for each encrypted message. The characters are sorted by frequency, and those with the same rank are paired to form an initial decryption key.

For example, if a ciphertext character is the most common character in the message, we initially map it to the most common character in the reference text.

This is only a first estimate. A message may use letters in different proportions from a long English novel, particularly when the message is short.

```python
from collections import Counter

alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

english_counts = Counter(moby_clean)
english_rank = sorted(alphabet,
                      key=lambda char: (-english_counts[char], alphabet.index(char)))

message = ciphertexts[2]
cipher_counts = Counter(message)
cipher_rank = sorted(alphabet,
                     key=lambda char: (-cipher_counts[char], alphabet.index(char)))

print("English character ranking:")
print(english_rank)
print("\nMessage 2 character ranking:")
print(cipher_rank)
```

```python
substitution_key = dict(zip(cipher_rank, english_rank))

print("Estimated substitution key for Message 2:")
for cipher_char, english_char in substitution_key.items():
    print(repr(cipher_char), "->", repr(english_char))
```

```python
def frequency_decrypt(message, key):
    decrypted = ""
    for char in message:
        decrypted += key.get(char, char)
    return decrypted

decrypted_message = frequency_decrypt(ciphertexts[2], substitution_key)

print("Message 2 - Frequency analysis result:")
print(decrypted_message)
```

```python
frequency_results = {}
frequency_keys = {}

for message_number in range(2, 13):
    message = ciphertexts[message_number]
    counts = Counter(message)

    ranked = sorted(alphabet,
                    key=lambda char: (-counts[char], alphabet.index(char)))
    key = dict(zip(ranked, english_rank))
    decrypted = frequency_decrypt(message, key)

    frequency_results[message_number] = decrypted
    frequency_keys[message_number] = key

    print("=" * 60)
    print("MESSAGE", message_number)
    print(decrypted[:400])
    print()
```

### Mathematical Explanation

If a character \(c\) appears \(N(c)\) times in a text containing \(N\) characters, its relative frequency is

$$
f(c)=\frac{N(c)}{N}.
$$

Substitution changes the characters but keeps their counts. If plaintext character \(x\) is replaced by ciphertext character \(y\), then

$$
N_{\mathrm{plaintext}}(x)=N_{\mathrm{ciphertext}}(y).
$$

We therefore match the frequency rankings from the reference text and the ciphertext to estimate a key. Characters with the same frequency are ordered alphabetically to keep the result reproducible. For a character that does not appear in a short message, its frequency is zero, so its assigned mapping is particularly uncertain.

### Results and Discussion

The frequency-based method gives an initial decryption for Messages 2–12. The Message 2 output contains a few recognisable English fragments, but much of the text is still incorrect. The character rankings alone do not identify all the substitutions.

The method is less dependable for short messages. Many characters appear only once or not at all, so their ranking can be quite different from that of the reference novel. There is also no check here for whether neighbouring letters form sensible words.

## 4. Manual Refinement of Substitution Ciphers

### Aim

The aim of this section is to improve the initial decryption obtained through character frequency analysis by manually adjusting the substitution key.

Although frequency analysis provides an initial estimate, it does not always correctly identify the original character mappings. This is because the frequency distribution of an individual message may differ from the distribution of the reference English text.

### Method

To improve the decryption, a Python function will be developed that allows two characters in the substitution key to exchange their roles.

After each swap, the encrypted message will be decrypted again using the updated key. The resulting text can then be examined to determine whether the change produces more recognizable English words and sentences.

By repeating this process, the substitution key can gradually be refined.

This method also provides the foundation for the Metropolis algorithm, which will automate the process of proposing and evaluating character swaps in Task 6.

```python

# Function to swap two characters in the substitution key

def swap_characters(key, char1, char2):

    # Copy the original key so it is not modified
    new_key = key.copy()

    # Find the encrypted characters mapped to char1 and char2
    cipher1 = next(c for c in new_key if new_key[c] == char1)
    cipher2 = next(c for c in new_key if new_key[c] == char2)

    # Exchange the two mappings
    new_key[cipher1], new_key[cipher2] = (
        new_key[cipher2], new_key[cipher1]
    )

    return new_key
```

### Mathematical Explanation

A substitution cipher can be represented by a permutation of the 27 characters in the alphabet.

Let \(K\) represent the current estimated decryption key.

A manual adjustment involves selecting two characters and exchanging their positions in the permutation. This operation is called a transposition.

If \(K'\) represents the modified key, then the new candidate plaintext is obtained by applying \(K'\) to the encrypted message.

The resulting plaintext is compared with the previous candidate to determine whether the substitution produces more meaningful English.

Because swapping two character assignments preserves the one-to-one nature of the substitution, the modified key remains a valid permutation.

However, manually testing different swaps becomes increasingly difficult as the number of possible substitutions grows. This motivates the use of an automated search algorithm in the later stages of the project.

## 5. Statistical Analysis Using Bigrams

### Aim

The aim of this section is to develop a statistical method for evaluating how closely a decrypted message resembles English text.

Character frequency analysis considers each letter independently, which does not account for the relationships between consecutive characters.

To improve this approach, we use bigrams, which are pairs of consecutive characters, such as TH, HE and IN.

### Method

We use the cleaned version of *Moby Dick* as a reference sample of English text.

First, we count the occurrences of each character and each possible bigram in the reference text.

These frequencies are then used to estimate the probability of one character following another.

Finally, we calculate a score for a candidate decrypted message based on these probabilities.

A higher score indicates that the candidate has a more plausible sequence of characters according to our statistical model of English.

```python

from collections import Counter

# Count individual characters in the reference text
character_counts = Counter(moby_clean)

# Count consecutive pairs of characters
bigram_counts = Counter()

for i in range(len(moby_clean) - 1):
    pair = moby_clean[i:i+2]
    bigram_counts[pair] += 1

# Display the 20 most common bigrams
print("20 most common bigrams:\n")

for pair, count in bigram_counts.most_common(20):
    print(repr(pair), ":", count)
```

```python

# Define the 27-character alphabet
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "

# Calculate probabilities for all possible bigrams
bigram_probabilities = {}

for char1 in alphabet:
    for char2 in alphabet:

        pair = char1 + char2

        probability = (
            bigram_counts[pair] + 1
        ) / character_counts[char1]

        bigram_probabilities[pair] = probability

print("Example bigram probabilities:\n")

for pair in ["TH", "HE", "IN", "ZZ", "QX"]:
    print(pair, ":", bigram_probabilities[pair])
```

```python

import math

# Function to calculate the statistical score of a message
def score_message(message):

    score = 0

    # Examine every consecutive pair of characters
    for i in range(len(message) - 1):

        pair = message[i:i+2]

        # Find the probability of this bigram
        probability = bigram_probabilities[pair]

        # Add its logarithm to the total score
        score += math.log(probability)

    return score
```

```python

# Example of normal English text
english_text = "THE QUICK BROWN FOX JUMPS OVER THE LAZY DOG"

# Example of random-looking text
random_text = "XQZ JKW PVB ZQX LMN WRT YKP FGH JZX"

# Calculate their scores
english_score = score_message(english_text)
random_score = score_message(random_text)

print("English text score:", english_score)
print("Random text score:", random_score)
```

## 6. Decryption Using the Metropolis Algorithm

### Aim

The aim of this section is to automate the process of improving substitution cipher keys using the Metropolis algorithm, a Monte Carlo method inspired by statistical physics.

In the previous section, we developed a statistical scoring function based on bigram probabilities. This function allows us to evaluate how closely a candidate decryption resembles English.

### Method

The algorithm begins with an estimated substitution key obtained through frequency analysis.

At each iteration, two randomly selected characters exchange their positions in the substitution key, producing a new candidate decryption.

The statistical score of the new message is compared with the current score.

If the new score is higher, the proposed change is accepted. If the score is lower, the change may still be accepted with a probability determined by the temperature parameter.

This allows the algorithm to explore different substitutions without becoming immediately trapped in a locally optimal solution.

The process is repeated many times, and the highest-scoring decryption is retained.

```python

import random
import math

def metropolis_decrypt(message, initial_key, steps=10000, T=1):

    # Start with the frequency-analysis key
    current_key = initial_key.copy()
    current_text = frequency_decrypt(message, current_key)
    current_score = score_message(current_text)

    # Store the best result found
    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    # Record the scores to analyse progress
    score_history = []

    for step in range(steps):

        # Randomly select two plaintext characters
        char1, char2 = random.sample(alphabet, 2)

        # Propose a new substitution key
        new_key = swap_characters(current_key, char1, char2)

        # Decrypt using the proposed key
        new_text = frequency_decrypt(message, new_key)

        # Calculate its score
        new_score = score_message(new_text)

        # Difference between the new and current scores
        difference = new_score - current_score

        # Metropolis acceptance rule
        if difference > 0 or random.random() < math.exp(difference / T):

            current_key = new_key
            current_text = new_text
            current_score = new_score

        # Update the best result
        if current_score > best_score:

            best_key = current_key.copy()
            best_text = current_text
            best_score = current_score

        # Record progress
        score_history.append(current_score)

    return best_text, best_key, best_score, score_history
```

### Mathematical Explanation

The Metropolis algorithm accepts proposed changes according to the probability:

$$
P = \min\left(1,e^{\Delta S/T}\right)
$$

where:

- \(\Delta S\) is the change in statistical score.
- \(T\) is the temperature parameter.
- \(P\) is the probability of accepting a proposed change.

At higher temperatures, the algorithm is more likely to accept changes that temporarily decrease the statistical score. This allows it to explore a wider range of substitution keys and potentially escape local optima.

In this experiment, T = 5 produced the highest statistical score and the most readable decryption of message 3. However, since the algorithm involves random changes, further independent simulations would be needed to determine whether this temperature consistently produces the best results.

Trigram analysis provides additional information about the structure of English compared with individual character frequencies and bigrams. For example, combinations such as "THE", "AND" and "ING" are common in English.

```python

from collections import Counter

# Count groups of three consecutive characters
trigram_counts = Counter()

for i in range(len(moby_clean) - 2):
    trigram = moby_clean[i:i+3]
    trigram_counts[trigram] += 1

# Calculate trigram probabilities with smoothing
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "
trigram_probabilities = {}

for a in alphabet:
    for b in alphabet:
        for c in alphabet:
            trigram = a + b + c
            prefix = a + b

            probability = (
                trigram_counts[trigram] + 1
            ) / (
                bigram_counts[prefix] + len(alphabet)
            )

            trigram_probabilities[trigram] = probability

print("20 most common trigrams:\n")

for trigram, count in trigram_counts.most_common(20):
    print(repr(trigram), ":", count)
```

```python

# ============================================================
# EXTENSION 9: IMPROVING THE DECRYPTION ALGORITHM
# Bigram, trigram and common-word scoring
# Metropolis algorithm with simulated annealing
# ============================================================

import os
import json
import math
import random
import re
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. LOAD THE REFERENCE TEXT AND ENCRYPTED MESSAGES
# ============================================================

alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "
N = len(alphabet)

# Load Moby Dick directly from the saved text file.
if os.path.exists("moby_clean.txt"):
    with open("moby_clean.txt", "r", encoding="utf-8") as f:
        moby_clean = f.read()
elif os.path.exists("moby.txt"):
    with open("moby.txt", "r", encoding="utf-8") as f:
        moby_clean = f.read()
else:
    raise FileNotFoundError(
        "Place moby_clean.txt or moby.txt in the Group Project folder."
    )

reference_text = moby_clean.upper()
reference_text = re.sub(r"[^A-Z ]", " ", reference_text)
reference_text = re.sub(r" +", " ", reference_text)

print("Reference text loaded successfully.")
print("Reference length:", len(reference_text))


# Load ciphertexts from the original notebook if necessary.
if "ciphertexts" not in globals() or not ciphertexts:

    filename = "code_breaking_ciphertexts.ipynb"

    if not os.path.exists(filename):
        raise FileNotFoundError(
            "Could not find code_breaking_ciphertexts.ipynb. "
            "Run the first ciphertext cell in Draft 2.ipynb "
            "or place the original notebook in this folder."
        )

    with open(filename, "r", encoding="utf-8") as f:
        notebook = json.load(f)

    ciphertexts = {}

    for cell in notebook["cells"]:
        source = "".join(cell.get("source", []))

        if cell["cell_type"] == "code" and "ciphertexts[1]" in source:
            exec(source, {"ciphertexts": ciphertexts})
            break

if not ciphertexts:
    raise ValueError("No ciphertexts were loaded.")

# Check that every ciphertext uses the expected alphabet.
for number, message in ciphertexts.items():
    invalid = set(message) - set(alphabet)
    if invalid:
        raise ValueError(
            f"Message {number} contains invalid characters: {invalid}"
        )

print("Encrypted messages loaded:", len(ciphertexts))


# ============================================================
# 2. CONSTRUCT BIGRAM AND TRIGRAM LANGUAGE MODELS
# ============================================================

character_counts = Counter(reference_text)

bigram_counts = Counter(
    reference_text[i:i+2]
    for i in range(len(reference_text)-1)
)

trigram_counts = Counter(
    reference_text[i:i+3]
    for i in range(len(reference_text)-2)
)

# Add-one smoothing avoids zero probabilities.
bigram_log = {}
trigram_log = {}

for a in alphabet:
    for b in alphabet:

        pair = a + b

        p2 = (
            (bigram_counts[pair] + 1)
            / (character_counts[a] + N)
        )

        bigram_log[pair] = math.log(p2)

        for c in alphabet:

            triple = a + b + c

            p3 = (
                (trigram_counts[triple] + 1)
                / (bigram_counts[pair] + N)
            )

            trigram_log[triple] = math.log(p3)

print("Bigram and trigram models constructed.")


# ============================================================
# 3. CONSTRUCT A COMMON ENGLISH WORD MODEL
# ============================================================

word_counts = Counter(reference_text.split())

common_words = {
    word: count
    for word, count in word_counts.items()
    if count >= 15 and 3 <= len(word) <= 12
}

word_weights = {
    word: math.log1p(count)
    for word, count in common_words.items()
}

print("Common English words identified:", len(common_words))


# ============================================================
# 4. DEFINE SCORING FUNCTIONS
# ============================================================

def score_bigrams(text):

    if len(text) < 2:
        return -100.0

    return sum(
        bigram_log[text[i:i+2]]
        for i in range(len(text)-1)
    ) / (len(text)-1)


def score_trigrams(text):

    if len(text) < 3:
        return -100.0

    return sum(
        trigram_log[text[i:i+3]]
        for i in range(len(text)-2)
    ) / (len(text)-2)


def score_words(text):

    if not text:
        return 0.0

    total = sum(
        word_weights.get(word, 0.0) * len(word)
        for word in text.split()
    )

    return total / len(text)


def improved_score(text):

    return (
        0.25 * score_bigrams(text)
        + 0.75 * score_trigrams(text)
        + 0.15 * score_words(text)
    )


# Test scoring functions.
english_test = "THE QUICK BROWN FOX JUMPS OVER THE LAZY DOG"
random_test = "XQZ JKW PVB ZQX LMN WRT YKP FGH BCD"

print("\nScoring test:")
print("English score:", round(improved_score(english_test), 4))
print("Random score:", round(improved_score(random_test), 4))


# ============================================================
# 5. INITIAL FREQUENCY-BASED SUBSTITUTION KEY
# ============================================================

def initial_frequency_key(ciphertext):

    cipher_counts = Counter(ciphertext)

    cipher_rank = sorted(
        alphabet,
        key=lambda c: (-cipher_counts[c], alphabet.index(c))
    )

    english_rank = sorted(
        alphabet,
        key=lambda c: (-character_counts[c], alphabet.index(c))
    )

    return dict(zip(cipher_rank, english_rank))


def decrypt_with_key(ciphertext, key):

    return "".join(key[c] for c in ciphertext)


def swap_key(key, a, b):

    new_key = key.copy()

    new_key[a], new_key[b] = new_key[b], new_key[a]

    return new_key


# ============================================================
# 6. METROPOLIS ALGORITHM WITH SIMULATED ANNEALING
# ============================================================

def annealing_decrypt(
    ciphertext,
    scoring_function,
    steps=15000,
    start_temperature=0.12,
    end_temperature=0.005,
    initial_key=None,
    seed=None
):

    rng = random.Random(seed)

    if initial_key is None:
        current_key = initial_frequency_key(ciphertext)
    else:
        current_key = initial_key.copy()

    current_text = decrypt_with_key(ciphertext, current_key)
    current_score = scoring_function(current_text)

    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    history = []
    accepted = 0

    for step in range(steps):

        fraction = step / max(steps-1, 1)

        T = start_temperature * (
            end_temperature / start_temperature
        ) ** fraction

        a, b = rng.sample(alphabet, 2)

        candidate_key = swap_key(current_key, a, b)

        candidate_text = decrypt_with_key(
            ciphertext, candidate_key
        )

        candidate_score = scoring_function(candidate_text)

        difference = candidate_score - current_score

        if difference >= 0 or rng.random() < math.exp(
            max(-700, difference / T)
        ):

            current_key = candidate_key
            current_text = candidate_text
            current_score = candidate_score
            accepted += 1

        if current_score > best_score:

            best_key = current_key.copy()
            best_text = current_text
            best_score = current_score

        if step % 100 == 0 or step == steps-1:
            history.append((step+1, best_score))

    return {
        "text": best_text,
        "key": best_key,
        "score": best_score,
        "history": history,
        "acceptance_rate": accepted / steps
    }


# ============================================================
# 7. MULTIPLE INDEPENDENT STARTING POINTS
# ============================================================

def multi_start_decrypt(
    ciphertext,
    scoring_function,
    runs=4,
    steps=15000,
    seed=42
):

    rng = random.Random(seed)

    best_result = None
    all_scores = []

    for run in range(runs):

        key = initial_frequency_key(ciphertext)

        if run > 0:

            for _ in range(10 + 5 * run):

                a, b = rng.sample(alphabet, 2)
                key = swap_key(key, a, b)

        result = annealing_decrypt(
            ciphertext,
            scoring_function,
            steps=steps,
            initial_key=key,
            seed=seed + run
        )

        all_scores.append(result["score"])

        if (
            best_result is None
            or result["score"] > best_result["score"]
        ):
            best_result = result

        print(
            f"Run {run+1}/{runs}: "
            f"score = {result['score']:.4f}"
        )

    best_result["all_scores"] = all_scores

    return best_result


# ============================================================
# 8. COMPARE ORIGINAL AND IMPROVED ALGORITHMS
# ============================================================

# Start with four messages.
# More messages can be added later.

messages_to_test = [2, 3, 4, 8]

STEPS = 15000
RUNS = 4

comparison_results = []
all_results = {}

for number in messages_to_test:

    ciphertext = ciphertexts[number]

    print("\n" + "=" * 65)
    print("MESSAGE", number)
    print("=" * 65)

    print("\nOriginal bigram scoring:")

    original = multi_start_decrypt(
        ciphertext,
        score_bigrams,
        runs=RUNS,
        steps=STEPS,
        seed=100 + number
    )

    print("\nImproved combined scoring:")

    improved = multi_start_decrypt(
        ciphertext,
        improved_score,
        runs=RUNS,
        steps=STEPS,
        seed=100 + number
    )

    all_results[number] = {
        "original": original,
        "improved": improved
    }

    print("\nORIGINAL DECRYPTION:")
    print(original["text"])

    print("\nIMPROVED DECRYPTION:")
    print(improved["text"])

    # Evaluate both texts using the same scoring model.
    original_eval = improved_score(original["text"])
    improved_eval = improved_score(improved["text"])

    comparison_results.append({
        "Message": number,
        "Length": len(ciphertext),
        "Original score": round(original_eval, 4),
        "Improved score": round(improved_eval, 4),
        "Difference": round(improved_eval-original_eval, 4)
    })


# ============================================================
# 9. DISPLAY COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(comparison_results)

print("\nCOMPARISON OF ORIGINAL AND IMPROVED METHODS")
display(results_df)

# These scores measure fit to the combined language model.
# They do not independently prove correct decryption.


# ============================================================
# 10. PLOT CONVERGENCE OF THE METROPOLIS ALGORITHM
# ============================================================

message_number = 8

original = all_results[message_number]["original"]
improved = all_results[message_number]["improved"]

x1, y1 = zip(*original["history"])
x2, y2 = zip(*improved["history"])

plt.figure(figsize=(10, 5))

plt.plot(
    x1,
    np.array(y1)-y1[0],
    label="Original bigram model"
)

plt.plot(
    x2,
    np.array(y2)-y2[0],
    label="Improved combined model"
)

plt.xlabel("Iteration")
plt.ylabel("Improvement in own objective score")

plt.title(
    f"Metropolis convergence for Message {message_number}"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 11. MEASURE RECOGNISED ENGLISH WORDS
# ============================================================

def recognised_word_fraction(text):

    words = [
        word for word in text.split()
        if len(word) >= 3
    ]

    if not words:
        return 0.0

    recognised = sum(
        word in common_words
        for word in words
    )

    return recognised / len(words)


quality_results = []

for number, result in all_results.items():

    original_text = result["original"]["text"]
    improved_text = result["improved"]["text"]

    quality_results.append({
        "Message": number,
        "Original recognised words (%)": round(
            100 * recognised_word_fraction(original_text), 1
        ),
        "Improved recognised words (%)": round(
            100 * recognised_word_fraction(improved_text), 1
        )
    })

quality_df = pd.DataFrame(quality_results)

print("\nRECOGNISED WORD COMPARISON")
display(quality_df)


# ============================================================
# 12. DISPLAY FINAL RESULTS AND SUBSTITUTION KEYS
# ============================================================

for number in messages_to_test:

    result = all_results[number]["improved"]

    print("\n" + "=" * 65)
    print(f"MESSAGE {number}: IMPROVED DECRYPTION")
    print("=" * 65)

    print("\nDecrypted text:")
    print(result["text"])

    print("\nRecovered substitution key:")

    for cipher_char in alphabet:

        print(
            f"{repr(cipher_char)} -> "
            f"{repr(result['key'][cipher_char])}"
        )

    print("\nFinal score:", round(result["score"], 4))

    print(
        "Acceptance rate:",
        round(result["acceptance_rate"]*100, 2),
        "%"
    )

print("\nEXTENSION 9 COMPLETED SUCCESSFULLY")
```

## Extension 9 – Results and Discussion

The aim of this extension was to investigate whether combining bigram and trigram statistics with common English word recognition could improve the performance of the Metropolis algorithm, particularly for messages that were difficult to decrypt using the original bigram-based approach.

### Results

Before introducing the improvements, we had successfully decrypted Messages **1, 2, 3, 4, 6 and 8** using the methods developed in the earlier stages of the project.

We initially tested the improved scoring function on Messages 2, 3, 4 and 8. Both the original bigram model and the improved combined model produced identical decryptions for these messages. This suggests that the original approach was already sufficiently effective for these particular ciphertexts, and the additional linguistic information did not produce a noticeable improvement in the final plaintext.

We then extended the investigation to the more challenging Messages 5, 6, 7, 9, 10, 11 and 12. For this experiment, we increased the number of iterations to 20,000 per run and performed five independent runs for each scoring method.

The results showed that the improved model was particularly useful for some of the previously unsolved messages.

**Message 5:** The original bigram model produced a partially understandable decryption containing several incorrect words. However, the improved model recovered a coherent English sentence:

*A very choice monkey was taken ill and refused to eat. It was thought, however, that its appetite might be stimulated by a pineapple.*

This demonstrates that incorporating trigram statistics and common English words can help the algorithm distinguish between plausible but incorrect substitutions and more meaningful plaintext.

**Message 7:** The original model failed to produce a meaningful sentence, whereas the improved model successfully recovered:

*From the beginning think what may be the end.*

This was an important result because Message 7 is relatively short, containing only 44 characters. Short ciphertexts provide limited statistical information, making frequency-based decryption more difficult. The successful recovery suggests that the additional linguistic information can be particularly beneficial when the available ciphertext is limited.

**Message 6:** Both models produced the same decryption, indicating that the original bigram-based approach was already effective for this message.

**Message 9:** Neither model successfully recovered the complete plaintext. However, the resulting text contained several recognisable French words and expressions, including *les*, *le*, *que*, *temps* and *conscience*. This suggests that Message 9 may be written in French rather than English.

**Messages 10, 11 and 12:** Both methods struggled to produce meaningful plaintext. These messages are extremely short, containing only 16, 12 and 31 characters respectively. Their limited length makes it difficult to estimate reliable language statistics and identify the correct substitution key.

### Discussion

Overall, the results suggest that the improved scoring function can offer advantages over the original bigram-only approach, particularly for shorter or more challenging English ciphertexts.

The most significant improvements were observed for Messages 5 and 7, which were successfully decrypted using the combined model after the original method failed to recover the correct plaintext.

However, the improvements were not consistent across all messages. For ciphertexts that were already successfully decrypted, both methods often produced identical results. Furthermore, the shortest messages remained difficult to solve, demonstrating the limitations of statistical frequency analysis when insufficient ciphertext is available.

An interesting observation was made for Message 9. Although the decryption was unsuccessful, the presence of recognisable French vocabulary suggests that the ciphertext may not be English. Since our language model was trained using *Moby Dick*, an English-language reference text, the scoring function favours English character combinations and words. Consequently, it may not be appropriate for decrypting messages written in other languages.

These results demonstrate that incorporating additional linguistic information into the Metropolis algorithm can improve its effectiveness, although performance remains dependent on ciphertext length and the language being analysed.

The results for Message 9 also highlight an important limitation of the current approach: its dependence on English-language statistics. ## Extension 10 – Adapting the Metropolis Algorithm to French Ciphertexts

### Motivation

In the previous extensions, we successfully decrypted Messages 1, 2, 3, 4, 5, 6, 7 and 8 using the Metropolis algorithm and improvements to the scoring function.

However, Message 9 remained unsolved. When examining the attempted decryption, we noticed several words and character combinations that appeared to resemble French rather than English. This raised the possibility that the message might have been encrypted from a French plaintext.

Until now, our algorithm has relied on statistical information extracted from *Moby Dick*, an English-language novel. This allowed us to construct bigram and trigram frequency models representing common patterns in English.

However, different languages have different letter frequencies and character combinations. For example, combinations that are common in French may be relatively uncommon in English. Therefore, an English-based scoring function may assign low scores to otherwise correct French plaintext.

### Aim

The aim of this extension is to investigate whether adapting the language model to French improves the Metropolis algorithm's ability to decrypt Message 9.

Rather than changing the underlying Metropolis algorithm, we will modify the statistical information used to evaluate candidate decryptions.

### Method

To construct a French-language model, we will use *Les Misérables* by Victor Hugo, a French novel first published in 1862.

The text will be processed in a similar way to *Moby Dick*, converting letters to uppercase, removing accents and retaining only the characters required by our substitution cipher.

From this cleaned text, we will calculate French bigram and trigram frequencies, allowing us to construct a new scoring function that favours character combinations commonly found in French.

We will then apply the Metropolis algorithm to Message 9 using both the original English-language model and the newly constructed French-language model.

To make the comparison fair, we will use the same number of iterations and independent runs for both models.

The resulting decryptions will be compared based on their readability, linguistic coherence and statistical scores. Since the models use different language distributions, their raw scores may not be directly comparable, so the quality of the recovered plaintext will be particularly important.

### Expected Outcome

We expect that, if Message 9 is written in French, the French-language model will be more effective at identifying the correct substitution key than the English-language model.

A successful decryption would provide evidence that the language of the reference text plays an important role in the performance of statistical cryptanalysis.

Even if the complete plaintext cannot be recovered, an improvement in the number of recognisable French words would suggest that the language model is better suited to this ciphertext.

This extension will therefore allow us to investigate the importance of language-specific statistical models and whether the Metropolis algorithm can be adapted to decrypt messages written in languages other than English.

```python

# ============================================================
# EXTENSION 10
# DECRYPTING MESSAGE 9 USING A FRENCH LANGUAGE MODEL
# Metropolis Algorithm with Bigrams, Trigrams and French Words
# ============================================================

import re
import math
import random
import string
from collections import Counter

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD THE CLEANED FRENCH NOVEL
# ============================================================

with open("les_miserables_clean.txt", "r", encoding="utf-8") as file:
    french_clean = file.read().upper()

french_clean = re.sub(r"[^A-Z ]", " ", french_clean)
french_clean = re.sub(r"\s+", " ", french_clean).strip()

print("French reference text loaded successfully.")
print("Reference length:", len(french_clean))


# ============================================================
# 2. LOAD MESSAGE 9
# ============================================================

# Uses the ciphertext dictionary created in the earlier cells.

if "ciphertexts" not in globals():
    raise ValueError(
        "The ciphertexts dictionary has not been loaded. "
        "Run your previous ciphertext-loading cell first."
    )

if 9 not in ciphertexts:
    raise ValueError("Message 9 was not found in ciphertexts.")

message9 = ciphertexts[9].upper()

message9 = re.sub(r"[^A-Z ]", " ", message9)
message9 = re.sub(r"\s+", " ", message9).strip()

print("\nMessage 9 loaded successfully.")
print("Message length:", len(message9))
print("\nEncrypted Message 9:")
print(message9)


# ============================================================
# 3. BUILD FRENCH BIGRAM AND TRIGRAM MODELS
# ============================================================

alphabet = string.ascii_uppercase + " "
N = len(alphabet)

# Count all overlapping bigrams and trigrams.
bigram_counts = Counter(
    french_clean[i:i+2]
    for i in range(len(french_clean) - 1)
)

trigram_counts = Counter(
    french_clean[i:i+3]
    for i in range(len(french_clean) - 2)
)

# Laplace smoothing prevents log(0).
alpha = 0.1

bigram_total = sum(bigram_counts.values())
trigram_total = sum(trigram_counts.values())

bigram_denominator = bigram_total + alpha * N**2
trigram_denominator = trigram_total + alpha * N**3

bigram_log = {
    a + b: math.log(
        (bigram_counts[a + b] + alpha) / bigram_denominator
    )
    for a in alphabet
    for b in alphabet
}

trigram_log = {
    a + b + c: math.log(
        (trigram_counts[a + b + c] + alpha)
        / trigram_denominator
    )
    for a in alphabet
    for b in alphabet
    for c in alphabet
}

print("\nFrench bigram and trigram models constructed.")
print("Unique bigrams:", len(bigram_counts))
print("Unique trigrams:", len(trigram_counts))


# ============================================================
# 4. IDENTIFY COMMON FRENCH WORDS
# ============================================================

# Count words appearing in the French novel.

word_counts = Counter(french_clean.split())

# Common French words are useful for identifying
# meaningful candidate decryptions.

common_french_words = {
    word
    for word, count in word_counts.items()
    if count >= 20 and len(word) >= 2
}

# Add common French function words.
common_french_words.update({
    "LE", "LA", "LES", "UN", "UNE", "DES",
    "DE", "DU", "ET", "EST", "SON", "SA",
    "SES", "QUE", "QUI", "QUOI", "DANS",
    "POUR", "AVEC", "SUR", "PAR", "PAS",
    "PLUS", "MAIS", "NOUS", "VOUS",
    "ILS", "ELLES", "ELLE", "IL",
    "JE", "TU", "AU", "AUX",
    "CE", "CET", "CETTE", "CES",
    "NE", "SE", "EN", "ON", "OU",
    "COMME", "TOUT", "TOUS",
    "BIEN", "FAIT", "TEMPS",
    "JOUR", "NUIT", "HOMME",
    "FEMME", "MONDE", "VIE"
})

print("\nFrench word model constructed.")
print("Recognised French words:", len(common_french_words))


# ============================================================
# 5. COMBINED FRENCH SCORING FUNCTION
# ============================================================

# Higher scores indicate more likely French plaintext.

BIGRAM_WEIGHT = 0.35
TRIGRAM_WEIGHT = 0.65
WORD_WEIGHT = 0.50

def french_score(text):

    text = text.upper()

    if len(text) < 3:
        return float("-inf")

    # Bigram score
    bigram_score = sum(
        bigram_log.get(
            text[i:i+2],
            math.log(alpha / bigram_denominator)
        )
        for i in range(len(text) - 1)
    )

    bigram_score /= max(1, len(text) - 1)

    # Trigram score
    trigram_score = sum(
        trigram_log.get(
            text[i:i+3],
            math.log(alpha / trigram_denominator)
        )
        for i in range(len(text) - 2)
    )

    trigram_score /= max(1, len(text) - 2)

    # French word recognition
    words = text.split()

    recognised = sum(
        len(word)
        for word in words
        if word in common_french_words
    )

    total_letters = sum(len(word) for word in words)

    word_score = (
        recognised / total_letters
        if total_letters > 0
        else 0
    )

    # Combined score
    score = (
        BIGRAM_WEIGHT * bigram_score
        + TRIGRAM_WEIGHT * trigram_score
        + WORD_WEIGHT * word_score
    )

    return score


# ============================================================
# 6. SUBSTITUTION CIPHER DECRYPTION
# ============================================================

# We use the same 27-character alphabet as before:
# A-Z and space.

def decrypt(ciphertext, key):

    translation = str.maketrans(
        alphabet,
        "".join(key)
    )

    return ciphertext.translate(translation)


# ============================================================
# 7. GENERATE A RANDOM SUBSTITUTION KEY
# ============================================================

def random_key():

    key = list(alphabet)
    random.shuffle(key)

    return key


# ============================================================
# 8. METROPOLIS ALGORITHM
# ============================================================

def metropolis_french(
    ciphertext,
    iterations=20000,
    temperature=1.0
):

    # Initial random substitution key
    current_key = random_key()

    current_text = decrypt(ciphertext, current_key)
    current_score = french_score(current_text)

    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    accepted = 0
    history = []

    for iteration in range(iterations):

        # Propose a new key by swapping two characters.
        proposed_key = current_key.copy()

        i, j = random.sample(range(N), 2)

        proposed_key[i], proposed_key[j] = (
            proposed_key[j],
            proposed_key[i]
        )

        proposed_text = decrypt(
            ciphertext,
            proposed_key
        )

        proposed_score = french_score(proposed_text)

        # Difference in scores
        delta = proposed_score - current_score

        # Metropolis acceptance rule
        # Scale normalised score differences by text length.
        log_acceptance = (
            delta * len(ciphertext) / temperature
        )

        if (
            log_acceptance >= 0
            or math.log(random.random()) < log_acceptance
        ):

            current_key = proposed_key
            current_text = proposed_text
            current_score = proposed_score

            accepted += 1

        # Save the best solution found.
        if current_score > best_score:

            best_score = current_score
            best_key = current_key.copy()
            best_text = current_text

        history.append(best_score)

    acceptance_rate = accepted / iterations

    return {
        "key": best_key,
        "plaintext": best_text,
        "score": best_score,
        "acceptance_rate": acceptance_rate,
        "history": history
    }


# ============================================================
# 9. RUN THE METROPOLIS ALGORITHM FIVE TIMES
# ============================================================

RUNS = 5
ITERATIONS = 20000

results = []

print("\n" + "=" * 65)
print("FRENCH METROPOLIS DECRYPTION - MESSAGE 9")
print("=" * 65)

for run in range(RUNS):

    result = metropolis_french(
        message9,
        iterations=ITERATIONS
    )

    results.append(result)

    print(
        f"Run {run + 1}/{RUNS}: "
        f"score = {result['score']:.4f}, "
        f"acceptance = "
        f"{result['acceptance_rate'] * 100:.2f}%"
    )


# ============================================================
# 10. SELECT THE BEST RESULT
# ============================================================

best_result = max(
    results,
    key=lambda result: result["score"]
)

print("\n" + "=" * 65)
print("BEST FRENCH DECRYPTION")
print("=" * 65)

print("\nEncrypted Message 9:")
print(message9)

print("\nBest French decryption:")
print(best_result["plaintext"])

print("\nFinal score:")
print(round(best_result["score"], 4))

print("\nAcceptance rate:")
print(
    round(best_result["acceptance_rate"] * 100, 2),
    "%"
)


# ============================================================
# 11. DISPLAY THE RECOVERED SUBSTITUTION KEY
# ============================================================

print("\nRecovered substitution key:")
print("(Cipher character -> Plaintext character)\n")

for cipher_char, plain_char in zip(
    alphabet,
    best_result["key"]
):

    print(
        f"{repr(cipher_char)} -> {repr(plain_char)}"
    )


# ============================================================
# 12. PLOT METROPOLIS CONVERGENCE
# ============================================================

plt.figure(figsize=(11, 5))

for run_number, result in enumerate(results, start=1):

    plt.plot(
        result["history"],
        label=f"Run {run_number}",
        alpha=0.8
    )

plt.xlabel("Iteration")
plt.ylabel("Best French language score")
plt.title(
    "Extension 10: French Metropolis Convergence "
    "for Message 9"
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 13. DISPLAY ALL FIVE DECRYPTIONS
# ============================================================

print("\n" + "=" * 65)
print("COMPARISON OF ALL FIVE RUNS")
print("=" * 65)

for run_number, result in enumerate(results, start=1):

    print(f"\nRUN {run_number}")
    print("-" * 50)

    print("Score:", round(result["score"], 4))

    print("Decryption:")
    print(result["plaintext"])


print("\nEXTENSION 10 EXPERIMENT COMPLETED SUCCESSFULLY")
```

### Extension 11 – Decrypting Message 12 Using a Perfect Pangram

After successfully decrypting Messages 10 and 11 through manual letter substitution, Message 12 was the only remaining encrypted message.

Upon examining the ciphertext, we noticed an interesting pattern: **it contains all 26 letters of the alphabet exactly once, along with 5 spaces**.

This suggests that the original plaintext could be a **perfect pangram**, which is a sentence containing every letter of the alphabet exactly once.

Furthermore, the message follows a specific word-length structure:

**6 – 5 – 3 – 4 – 3 – 5**

This means that the plaintext should consist of six words containing 6, 5, 3, 4, 3 and 5 letters respectively.

Since a substitution cipher preserves the number of letters and their positions, we can use this structure to narrow down the possible plaintexts.

Rather than relying on the Metropolis algorithm, we decided to use the assistance of artificial intelligence to search for perfect pangrams in languages that use the Latin alphabet.

Through this approach, we identified **one possible candidate** matching the required word-length pattern and containing all 26 letters exactly once.

Although this does not prove that the candidate is the original plaintext, it provides a promising solution that can be tested against Message 12 using a substitution key.

This final investigation demonstrates how recognising structural patterns and combining human reasoning with AI-assisted searching can provide an alternative approach when statistical decryption methods are unsuccessful.

```python

# ============================================================
# EXTENSION 11 - MESSAGE 12
# DECRYPTION USING A PERFECT PANGRAM
# ============================================================

from collections import Counter
import string

# Original encrypted Message 12
ciphertext_12 = ciphertexts[12]

# Proposed plaintext
candidate_12 = "QUARTZ GLYPH JOB VEXD CWM FINKS"

# ------------------------------------------------------------
# 1. CONSTRUCT THE SUBSTITUTION KEY
# ------------------------------------------------------------

# Match each ciphertext character with its corresponding
# character in the proposed plaintext.

assert len(ciphertext_12) == len(candidate_12), \
    "Ciphertext and plaintext lengths do not match."

key_12 = {}

for cipher_char, plain_char in zip(ciphertext_12, candidate_12):

    if cipher_char in key_12:
        assert key_12[cipher_char] == plain_char, \
            "Inconsistent substitution detected."
    else:
        key_12[cipher_char] = plain_char

# Verify that the substitution is one-to-one
assert len(set(key_12.values())) == len(key_12), \
    "The substitution key is not one-to-one."

# ------------------------------------------------------------
# 2. DISPLAY THE RECOVERED SUBSTITUTIONS
# ------------------------------------------------------------

print("MESSAGE 12 - RECOVERED SUBSTITUTIONS\n")

for cipher_char, plain_char in key_12.items():

    cipher_display = (
        "(space)" if cipher_char == " " else cipher_char
    )

    plain_display = (
        "(space)" if plain_char == " " else plain_char
    )

    print(f"{cipher_display} -> {plain_display}")

# ------------------------------------------------------------
# 3. DECRYPT MESSAGE 12
# ------------------------------------------------------------

decrypted_12 = "".join(
    key_12[char] for char in ciphertext_12
)

print("\n" + "=" * 60)
print("ORIGINAL ENCRYPTED MESSAGE:")
print(ciphertext_12)

print("\nDECRYPTED MESSAGE:")
print(decrypted_12)
print("=" * 60)

# ------------------------------------------------------------
# 4. VERIFY THE PERFECT PANGRAM
# ------------------------------------------------------------

letters = decrypted_12.replace(" ", "")
letter_counts = Counter(letters)

is_perfect_pangram = (
    len(letters) == 26
    and set(letters) == set(string.ascii_uppercase)
    and all(count == 1 for count in letter_counts.values())
)

word_lengths = [
    len(word) for word in decrypted_12.split()
]

print("\nPERFECT PANGRAM VERIFICATION")
print("Total letters:", len(letters))
print("Total spaces:", decrypted_12.count(" "))
print("Unique letters:", len(letter_counts))
print("Word-length pattern:", word_lengths)
print("Perfect pangram:", is_perfect_pangram)

print("\nEXTENSION 11 - MESSAGE 12 COMPLETED")
```
