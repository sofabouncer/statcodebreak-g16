## 4. Manual Refinement of the Substitution Key

### 4.1 Aim

Section 3 gave a first key for every message by matching the frequency ranks of the ciphertext with those of *Moby Dick*. For Message 2 the decryption starts with `ISERLOCV SOLMEI GREIERKED SNI CALM`. A few words are almost English (for example `CALM`, and `TSAT SE` further on looks like `THAT HE`), but many letters are still wrong. The reason is that a single passage does not have exactly the same letter frequencies as a whole novel, so letters with similar counts can end up in the wrong order.

In this section we write a small tool that lets the user swap two letters in the key and immediately see how the decryption changes. We then use it on Message 2, one swap at a time, until the passage can be read.

### 4.2 Method and Mathematical Background

The key for Message 2 is stored in the dictionary `frequency_keys[2]` from Section 3. Each dictionary key is a ciphertext character and its value is the plaintext character we think it stands for. All 27 characters (A–Z and space) appear exactly once as keys and exactly once as values, so the key is a permutation $\sigma$ of the 27 characters and the decryption of a ciphertext $c_1c_2\dots c_n$ is

$$
m_i = \sigma(c_i), \qquad i = 1, \dots, n .
$$

Swapping two plaintext characters $a$ and $b$ changes the key to

$$
\sigma' = \tau_{ab} \circ \sigma ,
$$

where $\tau_{ab}$ is the transposition that exchanges $a$ and $b$ and leaves every other character fixed. Since $\tau_{ab}$ and $\sigma$ are both bijections, $\sigma'$ is also a bijection, so after a swap we still have a valid substitution key. Also $\tau_{ab}\circ\tau_{ab}$ is the identity, so doing the same swap twice gives back the original key. We use this fact in the tests below.

We swap **plaintext** characters (the values of the dictionary) rather than ciphertext characters, because the user decides on a swap by reading the decrypted text. One swap changes every position in the message where these two letters appear, not only the word we were looking at. So before keeping a swap we check several words, not just one.

Design choices:

- `swap_characters` returns a new dictionary and does not change the key it was given. This keeps the Core 3 key for comparison, and if a guess turns out to be bad we can simply go back to the previous key.
- In the main experiment every swap starts from the **last accepted** key (continuous refinement). In Section 5 we also try some swaps that all start again from the Core 3 key (independent trials), and compare the two.
- After every accepted swap we save the key and the **full** decrypted text in two lists, so Section 5 can score exactly the texts we looked at here.

### 4.3 Implementation and Testing

The function below performs one swap. Inside the loop it reads the values from the old key `key` and writes to the copy `new_key`. If it changed the same dictionary it was reading, an entry changed from `A` to a space could later be changed back to `A`.

```python
def swap_characters(key, char1, char2):
    """
    Inputs: key - dictionary {ciphertext character: plaintext character}
            char1, char2 - the two plaintext characters to exchange
    Returns a new key in which char1 and char2 have swapped places.
    The key given as input is not changed.
    """
    # The two characters must be different and must both be plaintext values in the key
    assert char1 != char2, "Please choose two different characters."
    assert char1 in key.values() and char2 in key.values(), "Both characters must be plaintext values in the key."

    # Work on a copy, so the key we started from stays the same
    new_key = key.copy()

    # Read from the old key and write to the copy
    for cipher_char in key:
        if key[cipher_char] == char1:
            new_key[cipher_char] = char2
        elif key[cipher_char] == char2:
            new_key[cipher_char] = char1

    return new_key
```

The second test picks two random characters (as in the unit tests of Lecture 7.3), so each run tests a different swap on the real Core 3 key. We check that exactly two entries change and that swapping twice gives back the original key.

The input checks in `swap_characters` use `assert`. If a check fails, Python raises an `AssertionError` with our message (Lecture 12+.2, Method 1), so a wrong swap can never silently produce a broken key.

The next two functions let us look at a swap before we decide to keep it. `first_difference` finds the first position where two texts differ; we also use it again in Section 5. `test_swap` prints the same part of the passage before and after the swap, starting just before the first change, and returns the new key and the **whole** new decryption. It does not change the key it is given, so trying a swap is always safe.

```python
def first_difference(text1, text2):
    """
    Inputs: two strings of the same length
    Returns the first position where they differ, or -1 if they are the same
    """
    for i in range(len(text1)):
        if text1[i] != text2[i]:
            return i
    return -1


def test_swap(message, key, char1, char2):
    """
    Inputs: message - the ciphertext
            key - the key we start from
            char1, char2 - the plaintext characters to swap
    Prints the same part of the decryption before and after the swap.
    Returns the new key and the full new decryption.
    """
    new_key = swap_characters(key, char1, char2)
    before = frequency_decrypt(message, key)
    after = frequency_decrypt(message, new_key)

    # Start the preview 15 characters before the first change
    # (if nothing changes, first_difference gives -1 and we start at 0)
    start = max(0, first_difference(before, after) - 15)
    print("Before:", before[start:start + 90])
    print("After: ", after[start:start + 90])

    return new_key, after
```

**Typing a swap yourself.** The user can also type the two characters instead of editing the code. We follow the robust input functions of Lecture 12+.2: `input_swap_once` asks once and returns `None` if the input is wrong, and `input_swap` keeps asking until the input is correct. The checks inside `swap_characters` raise an `AssertionError`, which we catch with `try` and `except` so that a typing mistake does not stop the program. Because the space is hard to see when typed, the user types the word `SPACE` for it.

```python
def read_character(prompt):
    """
    Input: prompt - the text shown to the user
    Returns the character typed by the user in capitals ("SPACE" gives " ")
    """
    typed = input(prompt).upper()
    if typed == "SPACE":
        return " "
    return typed


def input_swap_once(key):
    """
    Input: key - the key the swap will be applied to
    Prompts the user for two plaintext characters.
    Returns the pair (char1, char2) if the input is correct, otherwise None.
    """
    try:
        char1 = read_character("First character (A-Z or SPACE): ")
        char2 = read_character("Second character (A-Z or SPACE): ")
        assert len(char1) == 1 and len(char2) == 1, "please type one character each time"
        # swap_characters checks the two characters and raises AssertionError if they are wrong
        swap_characters(key, char1, char2)
    except AssertionError as err:
        print("There was an input error: " + str(err))
        return None
    else:
        return (char1, char2)


def input_swap(key):
    """
    Input: key - the key the swap will be applied to
    Prompts the user until two correct characters are typed, then returns them.
    """
    outcome = None
    while outcome is None:
        outcome = input_swap_once(key)
    return outcome
```

It checks that each key is still a one-to-one key (27 entries and every character appears as a value), that the saved text really is the decryption of Message 2 with that key, and that from one step to the next only the two chosen characters have changed. The last output is the full raw decryption that Section 5 will score.

```python

# Check every saved step
assert len(history_keys) == len(accepted_swaps) + 1
assert history_keys[0] == cipher_dict

for step in range(len(history_keys)):
    key = history_keys[step]

    # Still a valid key: 27 entries and every character is used once as a value
    assert len(key) == 27
    for char in alphabet:
        assert char in key.values(), "A plaintext character is missing from the key"

    # The saved text is the decryption of Message 2 with this key
    assert len(history_texts[step]) == len(message_2)
    assert history_texts[step] == frequency_decrypt(message_2, key)

    # From step 1 on, only the two chosen characters may have changed
    if step > 0:
        char1 = accepted_swaps[step - 1][0]
        char2 = accepted_swaps[step - 1][1]
        old_key = history_keys[step - 1]
        for cipher_char in alphabet:
            if old_key[cipher_char] == char1:
                assert key[cipher_char] == char2
            elif old_key[cipher_char] == char2:
                assert key[cipher_char] == char1
            else:
                assert key[cipher_char] == old_key[cipher_char]

print(len(history_keys), "steps checked, each text has", len(message_2), "characters.")
print("\nFinal raw decryption of Message 2:\n")
print(history_texts[-1])
```

## 5. Scoring a Decryption with Bigrams

### 5.1 Aim

In Section 4 we decided between swaps by reading the text, for example S/H turns `TSE` into `THE` while E/T turns it into `EST`. This works for a long message but it is subjective and cannot be done automatically. The aim of this section is to write a function that gives any candidate decryption a number, so that a higher number means the text looks more like English. We then use it to score every step of the manual refinement in Section 4.

### 5.2 Method and Mathematical Background

We use the order of neighbouring characters. A **bigram** is a pair of characters next to each other, for example `TH`, `HE` or `E ` (E followed by a space). The space is treated like any other character. In ordinary English some pairs, like `TH`, are very common and others, like `QX`, almost never happen.

From the cleaned *Moby Dick* text of Section 2 we count

- $N(a)$: the number of times character $a$ appears,
- $N(ab)$: the number of times character $a$ is immediately followed by character $b$.

We move along the text one position at a time, so the string `AAA` contains two `AA` pairs. Following the project booklet, the probability that $a$ is followed by $b$ is estimated by

$$
p(a,b) = \frac{N(ab) + 1}{N(a)} ,
$$

and the score of a candidate message $m = m_1 m_2 \dots m_n$ is

$$
S(m) = \sum_{i=1}^{n-1} \log p(m_i, m_{i+1}) ,
$$

where $\log$ is the natural logarithm. A few remarks:

- The $+1$ in the numerator means that a pair which never appears in *Moby Dick* (for example two spaces in a row) still has $p > 0$, so its logarithm is finite. Because of this $+1$, the values $p(a,b)$ for a fixed $a$ do not add up exactly to 1, but this does not matter for comparing scores.
- Every $p(a,b)$ is less than 1, so every term is negative. A text with more common pairs has terms closer to 0 and therefore a **higher** (less negative) score.
- Taking logarithms turns a product of many small probabilities into a sum. Multiplying almost two thousand numbers smaller than 1 would give a number too small for the computer to store.
- A text of length $n$ has $n-1$ pairs, so the score depends on the length. We therefore only compare texts of the **same length**. All the decryptions of Message 2 have the same length, so they can be compared directly.

### 5.3 Implementation and Testing

Both counting the pairs in *Moby Dick* and scoring a message need the same step: move along a text and take each pair of neighbours. We write this step once, as a **generator function** (Lecture 12.2). `gen_bigrams(text)` does not build a list of all pairs. It gives them one at a time, when the `for` loop asks for the next one. For the reference text this matters, because a list of more than a million two-character strings would use tens of megabytes of memory. A generator is used up after one pass (Lecture 12.1), so we always call `gen_bigrams(text)` again instead of storing it.

```python
def gen_bigrams(text):
    """
    Input: text - a string
    Returns a generator that generates the neighbouring pairs text[0:2], text[1:3], ...
    (a text of length n gives n - 1 pairs)
    """
    for i in range(len(text) - 1):
        yield text[i:i + 2]


# UNIT TEST gen_bigrams HERE
pairs = []
for pair in gen_bigrams("THE "):
    pairs.append(pair)
print("Pairs in 'THE ':", pairs)
assert pairs == ["TH", "HE", "E "]

aaa_pairs = []
for pair in gen_bigrams("AAA"):
    aaa_pairs.append(pair)
assert aaa_pairs == ["AA", "AA"], "Overlapping pairs must both be counted"

empty_pairs = []
for pair in gen_bigrams("A"):
    empty_pairs.append(pair)
assert empty_pairs == [], "A single character has no pairs"
```

We count the pairs in *Moby Dick* only once and store $\log p(a,b)$ for all $27 \times 27 = 729$ pairs in a dictionary. Scoring a message then only needs one dictionary look-up per pair, and no counting or logarithms. This matters because the Metropolis algorithm in Section 6 will call the score function many thousands of times. (The running time is analysed at the end of this subsection.)

```python

from collections import Counter
import math

# Use the cleaned Moby Dick text from Section 2
moby_clean = cleaned_moby

# Count how often each single character appears in the cleaned Moby Dick text (Section 2)
character_counts = Counter(moby_clean)

# Start all 27 x 27 = 729 pairs at zero, so pairs that never appear are also in the table
bigram_counts = {}
for first in alphabet:
    for second in alphabet:
        bigram_counts[first + second] = 0

# Count each pair of neighbours (gen_bigrams moves along the text one position at a time)
for pair in gen_bigrams(moby_clean):
    bigram_counts[pair] = bigram_counts[pair] + 1

# Store log p(a,b) = log((N(ab) + 1) / N(a)) for every pair.
# The logarithms are worked out once here, not every time a message is scored.
bigram_log_weights = {}
for pair in bigram_counts:
    first = pair[0]
    probability = (bigram_counts[pair] + 1) / character_counts[first]
    bigram_log_weights[pair] = math.log(probability)

# Check the tables: every character is counted once, and a text of length L has L - 1 pairs
assert len(character_counts) == 27 and len(bigram_counts) == 729
assert sum(character_counts.values()) == len(moby_clean)
assert sum(bigram_counts.values()) == len(moby_clean) - 1

print("Characters in the reference text:", len(moby_clean))
print("Neighbouring pairs:", sum(bigram_counts.values()))
print("N(T) =", character_counts["T"], "  N(TH) =", bigram_counts["TH"])
print("p(T,H) =", (bigram_counts["TH"] + 1) / character_counts["T"])
```

**The score function.** Section 6 calls the score with one input only, `score_message(text)`, so we want to keep that form. But if `score_message` simply read the dictionary `bigram_log_weights` as a global variable, changing that variable later (for example in an extension with a different table) would silently change every score. Lecture 2.2 warns about exactly this. Instead we use a **closure** (Lecture 12+.1): `make_score_function` takes a table of log weights and returns a new function that scores a text with that table. The table is fixed at the moment the function is made, and Section 6 still calls `score_message(text)` in the same way.

The input text should only contain the 27 characters of the project (capital letters and spaces). Any other character is not in the table and gives a `KeyError`.

```python
def make_score_function(log_weights):
    """
    Input: log_weights - dictionary {pair: log p(a,b)} for all 729 pairs
    Returns: the closure score(message), which adds up log p(a,b)
             over all neighbouring pairs of message
    """
    def score(message):
        """
        Input: message - a candidate decryption (capital letters A-Z and spaces only)
        Returns the bigram score S(m). A higher score means more like English.
        """
        total = 0
        # A text of length n has n - 1 pairs, so "" and "A" have no pairs and score 0
        for pair in gen_bigrams(message):
            total = total + log_weights[pair]
        return total
    return score


# The score used in the rest of the project, built from the Moby Dick table
score_message = make_score_function(bigram_log_weights)
```

**Checking the score by hand.** We work out a few scores in a different way and compare. For `TH` there is one term and for `THE` there are two (`TH` and `HE`). Here we count the pairs with the string method `count`. This method does not count overlapping matches, but `TH`, `HE` and `" T"` cannot overlap with themselves, so the counts are correct. The pair of two spaces never appears in the cleaned book, so it checks the $+1$. The pair `AA` *can* overlap (`AAA` contains two), so for that one we count with our own loop, written without `gen_bigrams` so that the check does not depend on the code it is testing. Two decimal numbers are treated as equal if they differ by less than a tolerance, in the same way as in the Newton–Raphson and fixed-point examples (Lectures 3.3 and 12+.1).

```python
# UNIT TEST score_message HERE

TOLERANCE = 1e-9

def close(x, y):
    """
    Inputs: two floats x, y
    Returns True if they differ by less than TOLERANCE
    Global variable used: TOLERANCE
    """
    return abs(x - y) < TOLERANCE

p_th = (moby_clean.count("TH") + 1) / moby_clean.count("T")
p_he = (moby_clean.count("HE") + 1) / moby_clean.count("H")
p_space_t = (moby_clean.count(" T") + 1) / moby_clean.count(" ")

assert score_message("") == 0 and score_message("A") == 0             # no pairs at all
assert close(score_message("TH"), math.log(p_th))                       # one pair
assert close(score_message("THE"), math.log(p_th) + math.log(p_he))     # two pairs: TH and HE
assert close(score_message(" T"), math.log(p_space_t))                  # the space is a character too

# Two spaces never appear together in the cleaned book, so only the +1 is left
assert bigram_counts["  "] == 0
assert close(score_message("  "), math.log(1 / character_counts[" "]))

# AA can overlap, so we count it with a loop
aa_count = 0
for i in range(len(moby_clean) - 1):
    if moby_clean[i] == "A" and moby_clean[i + 1] == "A":
        aa_count = aa_count + 1
p_aa = (aa_count + 1) / character_counts["A"]
assert close(score_message("AAA"), 2 * math.log(p_aa))                  # AAA has two AA pairs

print("All hand-calculated scores agree with score_message.")
```

**Time and memory.** Following Lecture 11.3, let $L = 1{,}174{,}222$ be the length of the reference text and $n$ the length of the message.

- Building the tables goes through the reference text once and then through the 729 pairs once, so it takes time $\Theta(L)$. This is done only once.
- `score_message` does one dictionary look-up and one addition for each of the $n-1$ pairs, so it takes time $\Theta(n)$. It does not depend on $L$.
- If instead we counted the pairs in *Moby Dick* every time we scored a message, each score would take time $\Theta(L)$. For Message 2 that is about $1{,}174{,}222 / 1{,}946 \approx 600$ times more work per score, repeated thousands of times in Section 6.
- The stored table has only 729 entries, and `gen_bigrams` never stores all the pairs of the reference text at once.

To check the $\Theta(n)$ claim we use the doubling hypothesis of Lecture 11.3: for a linear algorithm, doubling the length should roughly double the time. We score the first 20,000, 40,000, 80,000 and 160,000 characters of the reference text, 20 times each, and look at the ratio of the times. The exact times depend on the computer and change slightly on each run, but the ratios should stay close to 2.

```python
# Doubling test (Lecture 11.3): for a Theta(n) function the ratio of times should be close to 2
from time import time

lengths = [20000, 40000, 80000, 160000]
times = []
for n in lengths:
    text = moby_clean[:n]
    start = time()
    for repeat in range(20):
        score_message(text)
    times.append((time() - start) / 20)      # average time for one score

print("  Length   Time (s)   Ratio to previous")
print(f"{lengths[0]:8}  {times[0]:9.5f}           -")
for i in range(1, len(lengths)):
    print(f"{lengths[i]:8}  {times[i]:9.5f}   {times[i] / times[i - 1]:9.2f}")
```

**Scores of the continuous refinement.** We need three tables below (the accepted steps, the rejected guess and the independent trials). They all have the same form: each row is a candidate text that is compared with another text. So we write one function `print_score_table` for all three. It takes the score function as an input, as in Lecture 12+.1, so the same table could later be printed with a different score (for example in an extension).

In the first table step 0 is the Core 3 decryption, and its score is the baseline score. "Change" compares each step with the step just before it. The last column shows a short piece of the text around the first position that changed, using `first_difference` from Section 4. All the texts have 1,946 characters, so their scores can be compared directly.

```python
def changed_part(before, after):
    """
    Inputs: two texts of the same length
    Returns a short piece around the first change, written as 'before -> after'
    """
    position = first_difference(before, after)
    if position == -1:
        return "(no change)"
    start = max(0, position - 5)
    return before[start:position + 10] + " -> " + after[start:position + 10]


def print_score_table(labels, texts, compare_texts, score_function):
    """
    Inputs: labels - the name of each row
            texts - the candidate decryptions
            compare_texts - for each row, the text it is compared with
            score_function - the function used to score a text
    Prints one row per text: label, length, score, change and the changed part.
    Returns the list of scores.
    """
    scores = []
    print("Swap  Length       Score      Change   Before -> after")
    for i in range(len(texts)):
        score = score_function(texts[i])
        change = score - score_function(compare_texts[i])
        excerpt = changed_part(compare_texts[i], texts[i])
        print(f"{labels[i]:4}  {len(texts[i]):6}  {score:10.3f}  {change:+10.3f}   {excerpt}")
        scores.append(score)
    return scores


# Row names: "base" for step 0, then the swap made at each step
step_labels = ["base"]
for pair in accepted_swaps:
    step_labels.append(pair[0] + "/" + pair[1])

# Each step is compared with the step before it (step 0 with itself)
previous_texts = [history_texts[0]] + history_texts[:-1]

step_scores = print_score_table(step_labels, history_texts, previous_texts, score_message)
baseline_score = step_scores[0]

# The I/T guess we rejected after step 4, compared with the step 4 text
print("\nRejected guess, compared with step 4:")
rejected_scores = print_score_table(["I/T"], [rejected_text], [history_texts[4]], score_message)
```

**Does the score agree with the manual choices?** Yes. Every one of the ten accepted swaps raised the score, from −6502.714 at the start to −4531.024 at the end. **Shuffled sentences.** In all three examples the real sentence scores higher than its shuffled version, although the two contain exactly the same characters. Part of the gap comes from double spaces in the shuffled versions: two spaces never appear together in the cleaned book, so each double space only gets the $+1$ and costs about $\log(1/219053) \approx -12.3$.

**Limitations.**

- The score only looks at pairs of neighbouring characters. It cannot tell whether a word is real or a sentence makes sense, and a high score does not prove that a decryption is correct or unique.
- The score depends on the length of the text, so we can only compare texts of the same length directly (or use the average per pair).
- A short message has only a few pairs. For Message 11 (12 characters, 11 pairs) many different keys will have similar scores, so the score gives much weaker evidence.
- The reference text is a 19th-century novel. Text in a different style, or in a different language, may get a low score even when it is decrypted correctly.

```python
# Recording the progress of the Metropolis algorithm
def metropolis_progress(message, initial_key, steps, T):

    # Start with the initial substitution key
    current_key = initial_key.copy()
    current_text = frequency_decrypt(message, current_key)
    current_score = score_message(current_text)

    # Store the best result found
    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    # Lists to record the scores throughout the algorithm
    current_scores = [current_score]
    best_scores = [best_score]

    # Repeat the algorithm for the specified number of iterations
    for i in range(steps):

        # Generate a new candidate key
        new_key = random_swap(current_key)

        # Decrypt the message and calculate its score
        new_text = frequency_decrypt(message, new_key)
        new_score = score_message(new_text)

        # Decide whether to accept the proposed swap
        if accept_swap(current_score, new_score, T):

            current_key = new_key
            current_text = new_text
            current_score = new_score

        # Update the best result if a higher score is found
        if current_score > best_score:

            best_key = current_key.copy()
            best_text = current_text
            best_score = current_score

        # Record the scores after each iteration
        current_scores.append(current_score)
        best_scores.append(best_score)

    return best_key, best_text, best_score, current_scores, best_scores
```

```python

# ============================================================
# EXTENSION 9: IMPROVING THE DECRYPTION ALGORITHM
# Bigram, trigram and common-word scoring
# Metropolis algorithm with simulated annealing
# ============================================================

import os
import json
import math
import random
import re
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# 1. LOAD THE REFERENCE TEXT AND ENCRYPTED MESSAGES
# ============================================================

alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "
N = len(alphabet)

# Load Moby Dick directly from the saved text file.
if os.path.exists("moby_clean.txt"):
    with open("moby_clean.txt", "r", encoding="utf-8") as f:
        moby_clean = f.read()
elif os.path.exists("moby.txt"):
    with open("moby.txt", "r", encoding="utf-8") as f:
        moby_clean = f.read()
else:
    raise FileNotFoundError(
        "Place moby_clean.txt or moby.txt in the Group Project folder."
    )

reference_text = moby_clean.upper()
reference_text = re.sub(r"[^A-Z ]", " ", reference_text)
reference_text = re.sub(r" +", " ", reference_text)

print("Reference text loaded successfully.")
print("Reference length:", len(reference_text))


# Load ciphertexts from the original notebook if necessary.
if "ciphertexts" not in globals() or not ciphertexts:

    filename = "code_breaking_ciphertexts.ipynb"

    if not os.path.exists(filename):
        raise FileNotFoundError(
            "Could not find code_breaking_ciphertexts.ipynb. "
            "Run the first ciphertext cell in Draft 2.ipynb "
            "or place the original notebook in this folder."
        )

    with open(filename, "r", encoding="utf-8") as f:
        notebook = json.load(f)

    ciphertexts = {}

    for cell in notebook["cells"]:
        source = "".join(cell.get("source", []))

        if cell["cell_type"] == "code" and "ciphertexts[1]" in source:
            exec(source, {"ciphertexts": ciphertexts})
            break

if not ciphertexts:
    raise ValueError("No ciphertexts were loaded.")

# Check that every ciphertext uses the expected alphabet.
for number, message in ciphertexts.items():
    invalid = set(message) - set(alphabet)
    if invalid:
        raise ValueError(
            f"Message {number} contains invalid characters: {invalid}"
        )

print("Encrypted messages loaded:", len(ciphertexts))


# ============================================================
# 2. CONSTRUCT BIGRAM AND TRIGRAM LANGUAGE MODELS
# ============================================================

character_counts = Counter(reference_text)

bigram_counts = Counter(
    reference_text[i:i+2]
    for i in range(len(reference_text)-1)
)

trigram_counts = Counter(
    reference_text[i:i+3]
    for i in range(len(reference_text)-2)
)

# Add-one smoothing avoids zero probabilities.
bigram_log = {}
trigram_log = {}

for a in alphabet:
    for b in alphabet:

        pair = a + b

        p2 = (
            (bigram_counts[pair] + 1)
            / (character_counts[a] + N)
        )

        bigram_log[pair] = math.log(p2)

        for c in alphabet:

            triple = a + b + c

            p3 = (
                (trigram_counts[triple] + 1)
                / (bigram_counts[pair] + N)
            )

            trigram_log[triple] = math.log(p3)

print("Bigram and trigram models constructed.")


# ============================================================
# 3. CONSTRUCT A COMMON ENGLISH WORD MODEL
# ============================================================

word_counts = Counter(reference_text.split())

common_words = {
    word: count
    for word, count in word_counts.items()
    if count >= 15 and 3 <= len(word) <= 12
}

word_weights = {
    word: math.log1p(count)
    for word, count in common_words.items()
}

print("Common English words identified:", len(common_words))


# ============================================================
# 4. DEFINE SCORING FUNCTIONS
# ============================================================

def score_bigrams(text):

    if len(text) < 2:
        return -100.0

    return sum(
        bigram_log[text[i:i+2]]
        for i in range(len(text)-1)
    ) / (len(text)-1)


def score_trigrams(text):

    if len(text) < 3:
        return -100.0

    return sum(
        trigram_log[text[i:i+3]]
        for i in range(len(text)-2)
    ) / (len(text)-2)


def score_words(text):

    if not text:
        return 0.0

    total = sum(
        word_weights.get(word, 0.0) * len(word)
        for word in text.split()
    )

    return total / len(text)


def improved_score(text):

    return (
        0.25 * score_bigrams(text)
        + 0.75 * score_trigrams(text)
        + 0.15 * score_words(text)
    )


# Test scoring functions.
english_test = "THE QUICK BROWN FOX JUMPS OVER THE LAZY DOG"
random_test = "XQZ JKW PVB ZQX LMN WRT YKP FGH BCD"

print("\nScoring test:")
print("English score:", round(improved_score(english_test), 4))
print("Random score:", round(improved_score(random_test), 4))


# ============================================================
# 5. INITIAL FREQUENCY-BASED SUBSTITUTION KEY
# ============================================================

def initial_frequency_key(ciphertext):

    cipher_counts = Counter(ciphertext)

    cipher_rank = sorted(
        alphabet,
        key=lambda c: (-cipher_counts[c], alphabet.index(c))
    )

    english_rank = sorted(
        alphabet,
        key=lambda c: (-character_counts[c], alphabet.index(c))
    )

    return dict(zip(cipher_rank, english_rank))


def decrypt_with_key(ciphertext, key):

    return "".join(key[c] for c in ciphertext)


def swap_key(key, a, b):

    new_key = key.copy()

    new_key[a], new_key[b] = new_key[b], new_key[a]

    return new_key


# ============================================================
# 6. METROPOLIS ALGORITHM WITH SIMULATED ANNEALING
# ============================================================

def annealing_decrypt(
    ciphertext,
    scoring_function,
    steps=15000,
    start_temperature=0.12,
    end_temperature=0.005,
    initial_key=None,
    seed=None
):

    rng = random.Random(seed)

    if initial_key is None:
        current_key = initial_frequency_key(ciphertext)
    else:
        current_key = initial_key.copy()

    current_text = decrypt_with_key(ciphertext, current_key)
    current_score = scoring_function(current_text)

    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    history = []
    accepted = 0

    for step in range(steps):

        fraction = step / max(steps-1, 1)

        T = start_temperature * (
            end_temperature / start_temperature
        ) ** fraction

        a, b = rng.sample(alphabet, 2)

        candidate_key = swap_key(current_key, a, b)

        candidate_text = decrypt_with_key(
            ciphertext, candidate_key
        )

        candidate_score = scoring_function(candidate_text)

        difference = candidate_score - current_score

        if difference >= 0 or rng.random() < math.exp(
            max(-700, difference / T)
        ):

            current_key = candidate_key
            current_text = candidate_text
            current_score = candidate_score
            accepted += 1

        if current_score > best_score:

            best_key = current_key.copy()
            best_text = current_text
            best_score = current_score

        if step % 100 == 0 or step == steps-1:
            history.append((step+1, best_score))

    return {
        "text": best_text,
        "key": best_key,
        "score": best_score,
        "history": history,
        "acceptance_rate": accepted / steps
    }


# ============================================================
# 7. MULTIPLE INDEPENDENT STARTING POINTS
# ============================================================

def multi_start_decrypt(
    ciphertext,
    scoring_function,
    runs=4,
    steps=15000,
    seed=42
):

    rng = random.Random(seed)

    best_result = None
    all_scores = []

    for run in range(runs):

        key = initial_frequency_key(ciphertext)

        if run > 0:

            for _ in range(10 + 5 * run):

                a, b = rng.sample(alphabet, 2)
                key = swap_key(key, a, b)

        result = annealing_decrypt(
            ciphertext,
            scoring_function,
            steps=steps,
            initial_key=key,
            seed=seed + run
        )

        all_scores.append(result["score"])

        if (
            best_result is None
            or result["score"] > best_result["score"]
        ):
            best_result = result

        print(
            f"Run {run+1}/{runs}: "
            f"score = {result['score']:.4f}"
        )

    best_result["all_scores"] = all_scores

    return best_result


# ============================================================
# 8. COMPARE ORIGINAL AND IMPROVED ALGORITHMS
# ============================================================

# Start with four messages.
# More messages can be added later.

messages_to_test = [2, 3, 4, 8]

STEPS = 15000
RUNS = 4

comparison_results = []
all_results = {}

for number in messages_to_test:

    ciphertext = ciphertexts[number]

    print("\n" + "=" * 65)
    print("MESSAGE", number)
    print("=" * 65)

    print("\nOriginal bigram scoring:")

    original = multi_start_decrypt(
        ciphertext,
        score_bigrams,
        runs=RUNS,
        steps=STEPS,
        seed=100 + number
    )

    print("\nImproved combined scoring:")

    improved = multi_start_decrypt(
        ciphertext,
        improved_score,
        runs=RUNS,
        steps=STEPS,
        seed=100 + number
    )

    all_results[number] = {
        "original": original,
        "improved": improved
    }

    print("\nORIGINAL DECRYPTION:")
    print(original["text"])

    print("\nIMPROVED DECRYPTION:")
    print(improved["text"])

    # Evaluate both texts using the same scoring model.
    original_eval = improved_score(original["text"])
    improved_eval = improved_score(improved["text"])

    comparison_results.append({
        "Message": number,
        "Length": len(ciphertext),
        "Original score": round(original_eval, 4),
        "Improved score": round(improved_eval, 4),
        "Difference": round(improved_eval-original_eval, 4)
    })


# ============================================================
# 9. DISPLAY COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(comparison_results)

print("\nCOMPARISON OF ORIGINAL AND IMPROVED METHODS")
display(results_df)

# These scores measure fit to the combined language model.
# They do not independently prove correct decryption.


# ============================================================
# 10. PLOT CONVERGENCE OF THE METROPOLIS ALGORITHM
# ============================================================

message_number = 8

original = all_results[message_number]["original"]
improved = all_results[message_number]["improved"]

x1, y1 = zip(*original["history"])
x2, y2 = zip(*improved["history"])

plt.figure(figsize=(10, 5))

plt.plot(
    x1,
    np.array(y1)-y1[0],
    label="Original bigram model"
)

plt.plot(
    x2,
    np.array(y2)-y2[0],
    label="Improved combined model"
)

plt.xlabel("Iteration")
plt.ylabel("Improvement in own objective score")

plt.title(
    f"Metropolis convergence for Message {message_number}"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 11. MEASURE RECOGNISED ENGLISH WORDS
# ============================================================

def recognised_word_fraction(text):

    words = [
        word for word in text.split()
        if len(word) >= 3
    ]

    if not words:
        return 0.0

    recognised = sum(
        word in common_words
        for word in words
    )

    return recognised / len(words)


quality_results = []

for number, result in all_results.items():

    original_text = result["original"]["text"]
    improved_text = result["improved"]["text"]

    quality_results.append({
        "Message": number,
        "Original recognised words (%)": round(
            100 * recognised_word_fraction(original_text), 1
        ),
        "Improved recognised words (%)": round(
            100 * recognised_word_fraction(improved_text), 1
        )
    })

quality_df = pd.DataFrame(quality_results)

print("\nRECOGNISED WORD COMPARISON")
display(quality_df)


# ============================================================
# 12. DISPLAY FINAL RESULTS AND SUBSTITUTION KEYS
# ============================================================

for number in messages_to_test:

    result = all_results[number]["improved"]

    print("\n" + "=" * 65)
    print(f"MESSAGE {number}: IMPROVED DECRYPTION")
    print("=" * 65)

    print("\nDecrypted text:")
    print(result["text"])

    print("\nRecovered substitution key:")

    for cipher_char in alphabet:

        print(
            f"{repr(cipher_char)} -> "
            f"{repr(result['key'][cipher_char])}"
        )

    print("\nFinal score:", round(result["score"], 4))

    print(
        "Acceptance rate:",
        round(result["acceptance_rate"]*100, 2),
        "%"
    )

print("\nEXTENSION 9 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# EXTENSION 10
# DECRYPTING MESSAGE 9 USING A FRENCH LANGUAGE MODEL
# Metropolis Algorithm with Bigrams, Trigrams and French Words
# ============================================================

import re
import math
import random
import string
from collections import Counter

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD THE CLEANED FRENCH NOVEL
# ============================================================

with open("les_miserables_clean.txt", "r", encoding="utf-8") as file:
    french_clean = file.read().upper()

french_clean = re.sub(r"[^A-Z ]", " ", french_clean)
french_clean = re.sub(r"\s+", " ", french_clean).strip()

print("French reference text loaded successfully.")
print("Reference length:", len(french_clean))


# ============================================================
# 2. LOAD MESSAGE 9
# ============================================================

# Uses the ciphertext dictionary created in the earlier cells.

if "ciphertexts" not in globals():
    raise ValueError(
        "The ciphertexts dictionary has not been loaded. "
        "Run your previous ciphertext-loading cell first."
    )

if 9 not in ciphertexts:
    raise ValueError("Message 9 was not found in ciphertexts.")

message9 = ciphertexts[9].upper()

message9 = re.sub(r"[^A-Z ]", " ", message9)
message9 = re.sub(r"\s+", " ", message9).strip()

print("\nMessage 9 loaded successfully.")
print("Message length:", len(message9))
print("\nEncrypted Message 9:")
print(message9)


# ============================================================
# 3. BUILD FRENCH BIGRAM AND TRIGRAM MODELS
# ============================================================

alphabet = string.ascii_uppercase + " "
N = len(alphabet)

# Count all overlapping bigrams and trigrams.
bigram_counts = Counter(
    french_clean[i:i+2]
    for i in range(len(french_clean) - 1)
)

trigram_counts = Counter(
    french_clean[i:i+3]
    for i in range(len(french_clean) - 2)
)

# Laplace smoothing prevents log(0).
alpha = 0.1

bigram_total = sum(bigram_counts.values())
trigram_total = sum(trigram_counts.values())

bigram_denominator = bigram_total + alpha * N**2
trigram_denominator = trigram_total + alpha * N**3

bigram_log = {
    a + b: math.log(
        (bigram_counts[a + b] + alpha) / bigram_denominator
    )
    for a in alphabet
    for b in alphabet
}

trigram_log = {
    a + b + c: math.log(
        (trigram_counts[a + b + c] + alpha)
        / trigram_denominator
    )
    for a in alphabet
    for b in alphabet
    for c in alphabet
}

print("\nFrench bigram and trigram models constructed.")
print("Unique bigrams:", len(bigram_counts))
print("Unique trigrams:", len(trigram_counts))


# ============================================================
# 4. IDENTIFY COMMON FRENCH WORDS
# ============================================================

# Count words appearing in the French novel.

word_counts = Counter(french_clean.split())

# Common French words are useful for identifying
# meaningful candidate decryptions.

common_french_words = {
    word
    for word, count in word_counts.items()
    if count >= 20 and len(word) >= 2
}

# Add common French function words.
common_french_words.update({
    "LE", "LA", "LES", "UN", "UNE", "DES",
    "DE", "DU", "ET", "EST", "SON", "SA",
    "SES", "QUE", "QUI", "QUOI", "DANS",
    "POUR", "AVEC", "SUR", "PAR", "PAS",
    "PLUS", "MAIS", "NOUS", "VOUS",
    "ILS", "ELLES", "ELLE", "IL",
    "JE", "TU", "AU", "AUX",
    "CE", "CET", "CETTE", "CES",
    "NE", "SE", "EN", "ON", "OU",
    "COMME", "TOUT", "TOUS",
    "BIEN", "FAIT", "TEMPS",
    "JOUR", "NUIT", "HOMME",
    "FEMME", "MONDE", "VIE"
})

print("\nFrench word model constructed.")
print("Recognised French words:", len(common_french_words))


# ============================================================
# 5. COMBINED FRENCH SCORING FUNCTION
# ============================================================

# Higher scores indicate more likely French plaintext.

BIGRAM_WEIGHT = 0.35
TRIGRAM_WEIGHT = 0.65
WORD_WEIGHT = 0.50

def french_score(text):

    text = text.upper()

    if len(text) < 3:
        return float("-inf")

    # Bigram score
    bigram_score = sum(
        bigram_log.get(
            text[i:i+2],
            math.log(alpha / bigram_denominator)
        )
        for i in range(len(text) - 1)
    )

    bigram_score /= max(1, len(text) - 1)

    # Trigram score
    trigram_score = sum(
        trigram_log.get(
            text[i:i+3],
            math.log(alpha / trigram_denominator)
        )
        for i in range(len(text) - 2)
    )

    trigram_score /= max(1, len(text) - 2)

    # French word recognition
    words = text.split()

    recognised = sum(
        len(word)
        for word in words
        if word in common_french_words
    )

    total_letters = sum(len(word) for word in words)

    word_score = (
        recognised / total_letters
        if total_letters > 0
        else 0
    )

    # Combined score
    score = (
        BIGRAM_WEIGHT * bigram_score
        + TRIGRAM_WEIGHT * trigram_score
        + WORD_WEIGHT * word_score
    )

    return score


# ============================================================
# 6. SUBSTITUTION CIPHER DECRYPTION
# ============================================================

# We use the same 27-character alphabet as before:
# A-Z and space.

def decrypt(ciphertext, key):

    translation = str.maketrans(
        alphabet,
        "".join(key)
    )

    return ciphertext.translate(translation)


# ============================================================
# 7. GENERATE A RANDOM SUBSTITUTION KEY
# ============================================================

def random_key():

    key = list(alphabet)
    random.shuffle(key)

    return key


# ============================================================
# 8. METROPOLIS ALGORITHM
# ============================================================

def metropolis_french(
    ciphertext,
    iterations=20000,
    temperature=1.0
):

    # Initial random substitution key
    current_key = random_key()

    current_text = decrypt(ciphertext, current_key)
    current_score = french_score(current_text)

    best_key = current_key.copy()
    best_text = current_text
    best_score = current_score

    accepted = 0
    history = []

    for iteration in range(iterations):

        # Propose a new key by swapping two characters.
        proposed_key = current_key.copy()

        i, j = random.sample(range(N), 2)

        proposed_key[i], proposed_key[j] = (
            proposed_key[j],
            proposed_key[i]
        )

        proposed_text = decrypt(
            ciphertext,
            proposed_key
        )

        proposed_score = french_score(proposed_text)

        # Difference in scores
        delta = proposed_score - current_score

        # Metropolis acceptance rule
        # Scale normalised score differences by text length.
        log_acceptance = (
            delta * len(ciphertext) / temperature
        )

        if (
            log_acceptance >= 0
            or math.log(random.random()) < log_acceptance
        ):

            current_key = proposed_key
            current_text = proposed_text
            current_score = proposed_score

            accepted += 1

        # Save the best solution found.
        if current_score > best_score:

            best_score = current_score
            best_key = current_key.copy()
            best_text = current_text

        history.append(best_score)

    acceptance_rate = accepted / iterations

    return {
        "key": best_key,
        "plaintext": best_text,
        "score": best_score,
        "acceptance_rate": acceptance_rate,
        "history": history
    }


# ============================================================
# 9. RUN THE METROPOLIS ALGORITHM FIVE TIMES
# ============================================================

RUNS = 5
ITERATIONS = 20000

results = []

print("\n" + "=" * 65)
print("FRENCH METROPOLIS DECRYPTION - MESSAGE 9")
print("=" * 65)

for run in range(RUNS):

    result = metropolis_french(
        message9,
        iterations=ITERATIONS
    )

    results.append(result)

    print(
        f"Run {run + 1}/{RUNS}: "
        f"score = {result['score']:.4f}, "
        f"acceptance = "
        f"{result['acceptance_rate'] * 100:.2f}%"
    )


# ============================================================
# 10. SELECT THE BEST RESULT
# ============================================================

best_result = max(
    results,
    key=lambda result: result["score"]
)

print("\n" + "=" * 65)
print("BEST FRENCH DECRYPTION")
print("=" * 65)

print("\nEncrypted Message 9:")
print(message9)

print("\nBest French decryption:")
print(best_result["plaintext"])

print("\nFinal score:")
print(round(best_result["score"], 4))

print("\nAcceptance rate:")
print(
    round(best_result["acceptance_rate"] * 100, 2),
    "%"
)


# ============================================================
# 11. DISPLAY THE RECOVERED SUBSTITUTION KEY
# ============================================================

print("\nRecovered substitution key:")
print("(Cipher character -> Plaintext character)\n")

for cipher_char, plain_char in zip(
    alphabet,
    best_result["key"]
):

    print(
        f"{repr(cipher_char)} -> {repr(plain_char)}"
    )


# ============================================================
# 12. PLOT METROPOLIS CONVERGENCE
# ============================================================

plt.figure(figsize=(11, 5))

for run_number, result in enumerate(results, start=1):

    plt.plot(
        result["history"],
        label=f"Run {run_number}",
        alpha=0.8
    )

plt.xlabel("Iteration")
plt.ylabel("Best French language score")
plt.title(
    "Extension 10: French Metropolis Convergence "
    "for Message 9"
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 13. DISPLAY ALL FIVE DECRYPTIONS
# ============================================================

print("\n" + "=" * 65)
print("COMPARISON OF ALL FIVE RUNS")
print("=" * 65)

for run_number, result in enumerate(results, start=1):

    print(f"\nRUN {run_number}")
    print("-" * 50)

    print("Score:", round(result["score"], 4))

    print("Decryption:")
    print(result["plaintext"])


print("\nEXTENSION 10 EXPERIMENT COMPLETED SUCCESSFULLY")
```

## Extension 11 – Manual Decryption of Messages 10, 11 and 12

After implementing the previous extensions, we successfully 
decrypted Messages 1–9. However, Messages 10, 11 and 12 
remain unsolved.

Due to the short length of these ciphertexts, the statistical 
methods used in the previous extensions were less effective, 
as there was insufficient information to reliably identify 
the correct substitution keys.

Therefore, in this final extension, we will adopt a more 
manual approach, experimenting with different letter 
substitutions and using human reasoning to recognise 
potential words and patterns.

Although most substitution attempts were unsuccessful, 
we will only present the substitutions that produced 
our most plausible decryptions.

The aim of this extension is to demonstrate how manual 
cryptanalysis and pattern recognition can complement 
automated statistical methods when dealing with short 
encrypted messages.

```python

# ============================================================
# EXTENSION 11 - MANUAL SUBSTITUTION DECRYPTION
# MESSAGES 10 AND 11
# ============================================================

# The ciphertexts are already stored in the dictionary
# 'ciphertexts' from the previous extensions.

# ------------------------------------------------------------
# MESSAGE 10
# ------------------------------------------------------------

key_10 = {
    'X': ' ',
    'G': 'M',
    'Y': 'E',
    'U': 'T',
    'P': 'A',
    'T': 'I',
    'B': 'D',
    'Q': 'N',
    'D': 'G',
    'F': 'H'
}

# ------------------------------------------------------------
# MESSAGE 11
# ------------------------------------------------------------

key_11 = {
    'F': ' ',
    'Z': 'S',
    'V': 'E',
    'E': 'Y',
    'K': 'O',
    'T': 'U',
    'O': 'N'
}

# ------------------------------------------------------------
# DECRYPTION FUNCTION
# ------------------------------------------------------------

def manual_decrypt(ciphertext, key):
    plaintext = ""

    for char in ciphertext.upper():
        if char in key:
            plaintext += key[char]
        else:
            plaintext += char

    return plaintext


# ------------------------------------------------------------
# APPLY SUBSTITUTIONS
# ------------------------------------------------------------

for number, key in [(10, key_10), (11, key_11)]:

    ciphertext = ciphertexts[number]
    plaintext = manual_decrypt(ciphertext, key)

    print("=" * 60)
    print(f"MESSAGE {number}")
    print("=" * 60)

    print("\nOriginal encrypted message:")
    print(ciphertext)

    print("\nRecovered substitutions:")
    for cipher_char, plain_char in key.items():
        if plain_char == " ":
            print(f"{cipher_char} -> (space)")
        else:
            print(f"{cipher_char} -> {plain_char}")

    print("\nDECRYPTED MESSAGE:")
    print(plaintext)

    print()


print("=" * 60)
print("EXTENSION 11 - MANUAL DECRYPTION COMPLETED")
print("=" * 60)
```

### Extension 11 – Decrypting Message 12 Using a Perfect Pangram

After successfully decrypting Messages 10 and 11 through manual letter substitution, Message 12 was the only remaining encrypted message.

Upon examining the ciphertext, we noticed an interesting pattern: **it contains all 26 letters of the alphabet exactly once, along with 5 spaces**.

This suggests that the original plaintext could be a **perfect pangram**, which is a sentence containing every letter of the alphabet exactly once.

Furthermore, the message follows a specific word-length structure:

**6 – 5 – 3 – 4 – 3 – 5**

This means that the plaintext should consist of six words containing 6, 5, 3, 4, 3 and 5 letters respectively.

Since a substitution cipher preserves the number of letters and their positions, we can use this structure to narrow down the possible plaintexts.

Rather than relying on the Metropolis algorithm, we decided to use the assistance of artificial intelligence to search for perfect pangrams in languages that use the Latin alphabet.

Through this approach, we identified **one possible candidate** matching the required word-length pattern and containing all 26 letters exactly once.

Although this does not prove that the candidate is the original plaintext, it provides a promising solution that can be tested against Message 12 using a substitution key.

This final investigation demonstrates how recognising structural patterns and combining human reasoning with AI-assisted searching can provide an alternative approach when statistical decryption methods are unsuccessful.

```python

# ============================================================
# EXTENSION 11 - MESSAGE 12
# DECRYPTION USING A PERFECT PANGRAM
# ============================================================

from collections import Counter
import string

# Original encrypted Message 12
ciphertext_12 = ciphertexts[12]

# Proposed plaintext
candidate_12 = "QUARTZ GLYPH JOB VEXD CWM FINKS"

# ------------------------------------------------------------
# 1. CONSTRUCT THE SUBSTITUTION KEY
# ------------------------------------------------------------

# Match each ciphertext character with its corresponding
# character in the proposed plaintext.

assert len(ciphertext_12) == len(candidate_12), \
    "Ciphertext and plaintext lengths do not match."

key_12 = {}

for cipher_char, plain_char in zip(ciphertext_12, candidate_12):

    if cipher_char in key_12:
        assert key_12[cipher_char] == plain_char, \
            "Inconsistent substitution detected."
    else:
        key_12[cipher_char] = plain_char

# Verify that the substitution is one-to-one
assert len(set(key_12.values())) == len(key_12), \
    "The substitution key is not one-to-one."

# ------------------------------------------------------------
# 2. DISPLAY THE RECOVERED SUBSTITUTIONS
# ------------------------------------------------------------

print("MESSAGE 12 - RECOVERED SUBSTITUTIONS\n")

for cipher_char, plain_char in key_12.items():

    cipher_display = (
        "(space)" if cipher_char == " " else cipher_char
    )

    plain_display = (
        "(space)" if plain_char == " " else plain_char
    )

    print(f"{cipher_display} -> {plain_display}")

# ------------------------------------------------------------
# 3. DECRYPT MESSAGE 12
# ------------------------------------------------------------

decrypted_12 = "".join(
    key_12[char] for char in ciphertext_12
)

print("\n" + "=" * 60)
print("ORIGINAL ENCRYPTED MESSAGE:")
print(ciphertext_12)

print("\nDECRYPTED MESSAGE:")
print(decrypted_12)
print("=" * 60)

# ------------------------------------------------------------
# 4. VERIFY THE PERFECT PANGRAM
# ------------------------------------------------------------

letters = decrypted_12.replace(" ", "")
letter_counts = Counter(letters)

is_perfect_pangram = (
    len(letters) == 26
    and set(letters) == set(string.ascii_uppercase)
    and all(count == 1 for count in letter_counts.values())
)

word_lengths = [
    len(word) for word in decrypted_12.split()
]

print("\nPERFECT PANGRAM VERIFICATION")
print("Total letters:", len(letters))
print("Total spaces:", decrypted_12.count(" "))
print("Unique letters:", len(letter_counts))
print("Word-length pattern:", word_lengths)
print("Perfect pangram:", is_perfect_pangram)

print("\nEXTENSION 11 - MESSAGE 12 COMPLETED")
```
